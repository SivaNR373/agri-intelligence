```python

from pyspark.sql.functions import *

# Landing location
landing_path = "/Volumes/workspace/agri_intelligence/landing/"

# Read CSV files
villages_df = spark.read.csv(
    landing_path + "villages.csv",
    header=True,
    inferSchema=False
)

farmers_df = spark.read.csv(
    landing_path + "farmers.csv",
    header=True,
    inferSchema=False
)

products_df = spark.read.csv(
    landing_path + "products.csv",
    header=True,
    inferSchema=False
)

bills_df = spark.read.csv(
    landing_path + "bills.csv",
    header=True,
    inferSchema=False
)

bill_items_df = spark.read.csv(
    landing_path + "bill_items.csv",
    header=True,
    inferSchema=False
)
```

```python
print("VILLAGES")
villages_df.printSchema()
villages_df.show(5)

print("FARMERS")
farmers_df.printSchema()
farmers_df.show(5)

print("PRODUCTS")
products_df.printSchema()
products_df.show(5)

print("BILLS")
bills_df.printSchema()
bills_df.show(5)

print("BILL ITEMS")
bill_items_df.printSchema()
bill_items_df.show(5)
```

```python
print("Villages:", villages_df.count())
print("Farmers:", farmers_df.count())
print("Products:", products_df.count())
print("Bills:", bills_df.count())
print("Bill Items:", bill_items_df.count())
```

```python
# Write Villages
villages_df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.agri_intelligence.bronze_villages")


# Write Farmers
farmers_df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.agri_intelligence.bronze_farmers")


# Write Products
products_df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.agri_intelligence.bronze_products")


# Write Bills
bills_df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.agri_intelligence.bronze_bills")


# Write Bill Items
bill_items_df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.agri_intelligence.bronze_bill_items")

print("All five Bronze tables created successfully!")
```

```python
print("BRONZE TABLE VALIDATION")

print("Villages:", spark.table("workspace.agri_intelligence.bronze_villages").count())

print("Farmers:", spark.table("workspace.agri_intelligence.bronze_farmers").count())

print("Products:", spark.table("workspace.agri_intelligence.bronze_products").count())

print("Bills:", spark.table("workspace.agri_intelligence.bronze_bills").count())

print("Bill Items:", spark.table("workspace.agri_intelligence.bronze_bill_items").count())
```
